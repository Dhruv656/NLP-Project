# Job Market NLP Analysis

This notebook prepares the merged job dataset for NLP-style analysis and creates three visualizations:
1. Top demanded skills
2. Salary analysis by job type
3. Location and remote/onsite hiring analysis

```python
import pandas as pd
import numpy as np
import re
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_style('whitegrid')

DATA_PATH = Path('master_jobs_dataset.csv')
df = pd.read_csv(DATA_PATH)
print(df.head())
print('\nShape:', df.shape)
```

```python
# Clean and normalize the dataset
for col in ['job_title', 'company', 'location', 'description', 'tags', 'category', 'workplace', 'department', 'type']:
    if col in df.columns:
        df[col] = df[col].fillna('').astype(str)

# Normalize job titles

def normalize_title(title):
    title = title.strip().lower()
    title = re.sub(r'\s+', ' ', title)
    title = title.replace('data science engineer', 'data scientist')
    title = title.replace('data scientist engineer', 'data scientist')
    title = title.replace('machine learning engineer', 'machine learning engineer')
    return title.title()


df['job_title_clean'] = df['job_title'].apply(normalize_title)

# Normalize salary values

def normalize_salary(x):
    if pd.isna(x):
        return np.nan
    s = str(x).strip()
    if not s:
        return np.nan
    s = s.replace('€', '$').replace('£', '$')
    s = re.sub(r'[^0-9\-.$]', ' ', s)
    s = re.sub(r'\s+', ' ', s).strip()
    return s


df['salary_clean'] = df['salary'].apply(normalize_salary)

# Create a simple salary numeric column when possible

def parse_salary_value(s):
    if pd.isna(s):
        return np.nan
    s = str(s)
    nums = re.findall(r'\d+', s)
    if not nums:
        return np.nan
    return float(nums[0]) * 1000 if len(nums) == 1 else float(nums[0])


df['salary_numeric'] = df['salary_clean'].apply(parse_salary_value)

# Create a basic category label from job title if missing
if 'category' in df.columns:
    df['category'] = df['category'].replace('', np.nan)
    df['category'] = df['category'].fillna(df['job_title_clean'])

print(df[['job_title', 'job_title_clean', 'salary', 'salary_clean']].head())
```

```python
# Prepare text for NLP-style analysis
# Tokenization example

def tokenize(text):
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    return re.split(r'\s+', text.strip())

for col in ['job_title_clean', 'description', 'tags']:
    df[col + '_tokens'] = df[col].apply(tokenize)

# Simple skill extraction from tags and text
skill_keywords = ['python', 'sql', 'aws', 'docker', 'nlp', 'tensorflow', 'pytorch', 'kubernetes', 'azure', 'spark']


def extract_skills(row):
    text = ' '.join(row['tags'].lower().split()) + ' ' + ' '.join(row['description'].lower().split())
    found = [skill for skill in skill_keywords if skill in text]
    return found


df['skills'] = df.apply(extract_skills, axis=1)

# Flatten extracted skills for counting
skill_series = df['skills'].explode()
skill_counts = skill_series.value_counts().head(10)
print(skill_counts)
```

```python
# Visualization 1: Top demanded skills
skill_counts = pd.Series([skill for skills in df['skills'] for skill in skills]).value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=skill_counts.values, y=skill_counts.index, palette='viridis')
plt.title('Top Demanded Skills')
plt.xlabel('Frequency')
plt.ylabel('Skill')
plt.tight_layout()
plt.show()
```

```python
# Visualization 2: Salary by job type
salary_plot = df.dropna(subset=['salary_numeric']).copy()
if 'category' in salary_plot.columns and not salary_plot.empty:
    salary_plot['category_short'] = salary_plot['category'].apply(lambda x: str(x)[:25])
    grouped = salary_plot.groupby('category_short')['salary_numeric'].mean().sort_values(ascending=False).head(10)
    plt.figure(figsize=(10, 5))
    sns.barplot(x=grouped.values, y=grouped.index, palette='magma')
    plt.title('Average Salary by Job Category')
    plt.xlabel('Average Salary (approx.)')
    plt.ylabel('Category')
    plt.tight_layout()
    plt.show()
else:
    print('No salary data available for plotting.')
```

```python
# Visualization 3: Hiring by location and remote/onsite ratio
# Create a simple location summary

df['location_clean'] = df['location'].fillna('').astype(str)
df['is_remote'] = df['location_clean'].str.lower().str.contains('remote|worldwide|anywhere|hybrid', na=False)

location_counts = df['location_clean'].str.split(',').str[0].str.strip()
location_counts = location_counts[location_counts != ''].value_counts().head(10)

plt.figure(figsize=(10, 5))
sns.barplot(x=location_counts.values, y=location_counts.index, palette='cividis')
plt.title('Top Hiring Locations')
plt.xlabel('Number of Postings')
plt.ylabel('Location')
plt.tight_layout()
plt.show()

remote_ratio = df['is_remote'].mean() * 100
print(f'Remote-related postings: {remote_ratio:.1f}%')

plt.figure(figsize=(6, 4))
plt.pie([remote_ratio, 100 - remote_ratio], labels=['Remote/Hybrid', 'Other'], autopct='%1.1f%%', startangle=90, colors=['#4C78A8', '#F58518'])
plt.title('Remote vs Other Work Mode')
plt.axis('equal')
plt.show()
```

## NLP Preprocessing, Skill Extraction, and Market Insights

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report
import warnings
warnings.filterwarnings('ignore')

# Stopword removal + simple lemmatization-like normalization
stop_words = set([
    'the','and','for','with','in','on','of','to','a','an','is','are','be','this','that','our','you','your','will','work','jobs','job','remote','team','company','role','skills','experience','using','developing','develop','data','science','engineer','engineering','software','developer','analyst','manager','product','business','technical','technical','senior','junior','lead','principal'
])


def normalize_text(text):
    text = str(text).lower()
    text = re.sub(r'[^a-z0-9\s]', ' ', text)
    tokens = [t for t in re.split(r'\s+', text.strip()) if t and t not in stop_words]
    return ' '.join(tokens)


df['text_for_nlp'] = df[['job_title_clean', 'description', 'tags']].astype(str).agg(lambda x: ' '.join(x), axis=1)
df['clean_text'] = df['text_for_nlp'].apply(normalize_text)
print(df[['job_title_clean', 'clean_text']].head())
```

```python
# Task 1: Data cleaning and normalization summary
print('Rows after cleaning:', df.shape[0])
print('Missing job_title:', df['job_title_clean'].isna().sum())
print('Missing salary:', df['salary_numeric'].isna().sum())
```

```python
# Task 2: Tokenization example
sample_tokens = tokenize(df['job_title_clean'].iloc[0])
print('Tokenized sample:', sample_tokens)
```

```python
# Task 3: Stopword removal and lemmatization-like normalization
print('Cleaned sample text:', df['clean_text'].iloc[0])
```

```python
# Task 4: Skill extraction
skill_keywords = ['python', 'sql', 'aws', 'docker', 'nlp', 'tensorflow', 'pytorch', 'kubernetes', 'azure', 'spark']

def count_skills(text):
    text = str(text).lower()
    return {skill: text.count(skill) for skill in skill_keywords}

skill_df = pd.DataFrame([count_skills(t) for t in df['clean_text']])
skill_summary = skill_df.sum().sort_values(ascending=False).head(10)
print('Top skill counts:\n', skill_summary)
```

```python
# Task 5: Keyword frequency analysis
word_counts = pd.Series(' '.join(df['clean_text']).split()).value_counts().head(20)
print('Top words:\n', word_counts)
```

```python
# Task 6: Job clustering with TF-IDF and KMeans
vectorizer = TfidfVectorizer(max_features=300)
X = vectorizer.fit_transform(df['clean_text'])

k = 5
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
clusters = kmeans.fit_predict(X)
df['cluster'] = clusters
print('Silhouette score:', silhouette_score(X, clusters))
print(df['cluster'].value_counts())
```

```python
# Task 7: Salary analysis
salary_summary = df.dropna(subset=['salary_numeric']).groupby('job_title_clean')['salary_numeric'].mean().sort_values(ascending=False).head(10)
print('Top salary roles:\n', salary_summary)
```

```python
# Task 8: Location analysis
location_counts = df['location'].fillna('').astype(str).str.split(',').str[0].str.strip()
location_counts = location_counts[location_counts != ''].value_counts().head(10)
print('Top locations:\n', location_counts)
remote_ratio = df['is_remote'].mean() * 100
print('Remote-related ratio:', round(remote_ratio, 2), '%')
```

```python
# Task 9: Tag analysis
all_tags = []
for tags in df['tags']:
    all_tags.extend([t.strip().lower() for t in str(tags).split(',') if t.strip()])
print('Top tags:', pd.Series(all_tags).value_counts().head(10))
```

```python
# Task 10: Job title classification
def classify_title(title):
    t = str(title).lower()
    if any(k in t for k in ['data scientist', 'machine learning', 'ai', 'nlp', 'deep learning']):
        return 'AI/ML'
    if any(k in t for k in ['devops', 'cloud', 'infrastructure', 'site reliability', 'platform']):
        return 'DevOps'
    if any(k in t for k in ['software', 'developer', 'engineer', 'programmer']):
        return 'Software Engineering'
    if any(k in t for k in ['product', 'manager', 'leadership', 'business']):
        return 'Product/Management'
    if any(k in t for k in ['analyst', 'business', 'operations']):
        return 'Business/Analyst'
    return 'Other'


df['predicted_category'] = df['job_title_clean'].apply(classify_title)
print(df['predicted_category'].value_counts())
```

```python
# Task 11: Sentiment analysis (simple rule-based)
positive_words = ['growth', 'opportunity', 'remote', 'flexible', 'innovative', 'learn', 'benefit', 'great', 'excellent']
demanding_words = ['urgent', 'immediately', 'must', 'required', 'strict', 'deadline', 'fast-paced']

def sentiment_score(text):
    t = str(text).lower()
    pos = sum(1 for w in positive_words if w in t)
    dem = sum(1 for w in demanding_words if w in t)
    if pos > dem:
        return 'positive'
    if dem > pos:
        return 'demanding'
    return 'neutral'


df['sentiment'] = df['description'].apply(sentiment_score)
print(df['sentiment'].value_counts())
```

```python
# Task 12: TF-IDF analysis by job type
vectorizer_tfidf = TfidfVectorizer(stop_words='english', ngram_range=(1,2), max_features=100)
X_tfidf = vectorizer_tfidf.fit_transform(df['clean_text'])
print('TF-IDF vocabulary sample:', list(vectorizer_tfidf.vocabulary_.keys())[:20])
```

```python
# Task 13: Final summary insights
print('Most common skill:', skill_summary.index[0] if not skill_summary.empty else 'N/A')
print('Highest salary role:', salary_summary.index[0] if not salary_summary.empty else 'N/A')
print('Top location:', location_counts.index[0] if not location_counts.empty else 'N/A')
print('Remote ratio:', round(remote_ratio, 2), '%')
```
